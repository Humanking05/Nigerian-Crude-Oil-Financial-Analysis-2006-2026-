Rationale for Financial Analysis: Nigerian Crude Oil Dataset

1. Why We are Using This Dataset
Crude oil is the cornerstone of the Nigerian economy, accounting for the vast majority of foreign exchange earnings and a significant portion of the national budget. This specific dataset provides a granular monthly view of:

Price Volatility: Global market price fluctuations.

  * Operational Output: Volume of oil produced domestically (domProd).

  * Trade Dynamics: The volume of oil sent to the international market (crudeOilExp).

  * By using this data, we are analyzing the primary engine of Nigeria's macroeconomic stability.

By using this data, we are analyzing the primary engine of Nigeria's macroeconomic stability.

2. The Purpose of the Analysis

The primary goal is to transform raw production and price figures into actionable financial insights. Specifically, we aim to:

  * Revenue Modeling: Estimate gross export revenues by combining price and export volume.

  * Trend Identification: Spot cyclical patterns in production and identify periods of significant growth or decline.

  * Correlation Assessment: Determine how domestic production levels respond to global price changes.

  * Forecasting: Evaluate the projected data (up to 2026) to understand future fiscal expectations and potential budget deficits or surpluses.

3. Why This is Beneficial

Performing this analysis offers several strategic advantages:

  * Informed Decision Making: Provides a data-backed foundation for discussing fiscal policy, subsidies, and infrastructure investment.

  * Risk Management: Helps identify the economy's sensitivity to "price shocks" in the global market.

  * Efficiency Tracking: By comparing domProd (Total Production) vs crudeOilExp (Exports), we can infer domestic consumption patterns and the efficiency of the value chain.

  * Economic Predictability: For investors and policy-makers, understanding these trends reduces uncertainty regarding Nigeria's future dollar liquidity and sovereign credit strength.

```python
# import packages
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# import the dataset
df = pd.read_csv(r"C:\Users\ACER\Desktop\Portfolios\crudeoil\Crude_Oil_Data.csv")
```

```python
# check few rows of the dataset
df.head()
```

```python
# Transform 'period' to datetime format
df['period'] = pd.to_datetime(df['period'])
df.head()
```

```python
# Calculate daily export revenue in millions of USD
df['daily_rev'] = df['crudeOilPrice'] * df['crudeOilExp']

print(f"Mean Daily Revenue: ${df['daily_rev'].mean():.2f}M")
print(f"Max Daily Revenue: ${df['daily_rev'].max():.2f}M")
print(f"Min Daily Revenue: ${df['daily_rev'].min():.2f}M")
```

```python
# Financial risk metrics
price_std = df['crudeOilPrice'].std()
price_mean = df['crudeOilPrice'].mean()
coeff_of_variation = (price_std / price_mean) * 100

print(f"Price Volatility (Std Dev): ${price_std:.2f}")
print(f"Risk Coefficient (CV): {coeff_of_variation:.2f}%")
```

```python
# Calculate the percentage of production that is actually exported
df['export_ratio'] = (df['crudeOilExp'] / df['domProd']) * 100

print(f"Average Export Efficiency: {df['export_ratio'].mean():.2f}%")
```

```python
# Aggregate revenue by year and calculate growth rate
yearly_rev = df.groupby('tyear')['daily_rev'].mean()
yoy_growth = yearly_rev.pct_change() * 100

print("Yearly Revenue Growth Rates (%):")
print(yoy_growth.tail())
```

```python
# Calculating 12-month Moving Averages for smoothing
df['price_sma12'] = df['crudeOilPrice'].rolling(window=12).mean()
df['prod_sma12'] = df['domProd'].rolling(window=12).mean()

# Check latest trend direction
print(df[['period', 'price_sma12', 'prod_sma12']].tail())
```

```python
# Seasonal Revenue Cycle
df['revenue'] = df['crudeOilPrice'] * df['crudeOilExp']

# Group by month to find seasonal performance
seasonal_performance = df.groupby('tmonth')['revenue'].mean().sort_values(ascending=False)

print("Average Daily Revenue by Month:")
print(seasonal_performance)
```

```python
# Comparing early 5-year average vs last 5-year average production
early_prod = df[df['tyear'] <= 2010]['domProd'].mean()
recent_prod = df[df['tyear'] >= 2022]['domProd'].mean()

print(f"Structural Decline: {((recent_prod - early_prod) / early_prod) * 100:.2f}%")
```

```python
df = df.sort_values('period')
df['monthly_rev_est'] = df['crudeOilPrice'] * df['crudeOilExp'] * 30 # Approx 30 days

# Calculate Cumulative Revenue
df['cumulative_wealth'] = df['monthly_rev_est'].cumsum()

print(f"Total Wealth Generated (Estimated): ${df['cumulative_wealth'].iloc[-1] / 1000:.2f} Billion")
```

```python
avg_p_2026 = df[df['tyear'] == 2026]['crudeOilPrice'].mean()
avg_e_2026 = df[df['tyear'] == 2026]['crudeOilExp'].mean()

# Calculate impact of +/- $10 price change
price_impact_minus = (avg_p_2026 - 10) * avg_e_2026
price_impact_plus = (avg_p_2026 + 10) * avg_e_2026

print(f"Base 2026 Daily Revenue: ${avg_p_2026 * avg_e_2026:.2f}M")
print(f"Revenue at -$10 Price: ${price_impact_minus:.2f}M")
print(f"Revenue at +$10 Price: ${price_impact_plus:.2f}M")
```

```python
avg_p_2026 = df[df['tyear'] == 2026]['crudeOilPrice'].mean()
avg_e_2026 = df[df['tyear'] == 2026]['crudeOilExp'].mean()

# Calculate impact of +/- 10% volume change
vol_impact_minus = avg_p_2026 * (avg_e_2026 * 0.90)
vol_impact_plus = avg_p_2026 * (avg_e_2026 * 1.10)

print(f"Revenue at -10% Volume: ${vol_impact_minus:.2f}M")
print(f"Revenue at +10% Volume: ${vol_impact_plus:.2f}M")
```

```python
avg_p_2026 = df[df['tyear'] == 2026]['crudeOilPrice'].mean()
avg_e_2026 = df[df['tyear'] == 2026]['crudeOilExp'].mean()

# Combined shock: -$10 price AND -10% volume
shock_revenue = (avg_p_2026 - 10) * (avg_e_2026 * 0.90)
base_revenue = avg_p_2026 * avg_e_2026
pct_loss = ((shock_revenue - base_revenue) / base_revenue) * 100

print(f"Shock Scenario Revenue: ${shock_revenue:.2f}M")
print(f"Total Fiscal Loss: {pct_loss:.2f}%")
```

```python
avg_e_2026 = df[df['tyear'] == 2026]['crudeOilExp'].mean()

# Minimum revenue threshold
threshold = 50.0 
break_even_price = threshold / avg_e_2026

print(f"Minimum Price to maintain $50M/day: ${break_even_price:.2f}")
```

### Summary of the 'Structural Decline' Risk

The Nigerian Crude Oil industry is facing significant structural decline, as evidenced by a 42.52% decrease in production from pre-2010 levels to post-2022. This decline can be attributed to several factors including:

1. **Reserves Depletion**: Natural depletion of oil reserves over time.
2. **Technological Obsolescence**: The industry may have lagged behind in adopting new, more efficient drilling and extraction technologies.
3. **Infrastructure Aging**: Older infrastructure and production facilities are less productive than modern ones.
4. **Environmental Regulations**: Stringent environmental regulations can limit the ability to extract oil from certain areas.

The structural decline poses a significant risk for Nigeria's economy, which heavily relies on crude oil exports for revenue generation. If left unaddressed, this trend could lead to reduced government revenues and slower economic growth.

### Implications of the 'Shock Scenario' on National Budgeting

A shock scenario involving a $10 price drop along with a 10% volume reduction results in a substantial fiscal loss of 22.82%. This implies that:

- **Revenue Volatility**: The national budget will face significant unpredictability and instability due to the high volatility in oil prices.
- **Fiscal Risk Management**: The government must develop robust mechanisms to mitigate this risk, such as diversifying the economy and establishing a sovereign wealth fund to buffer against price fluctuations.
- **Emergency Funds**: Increased allocation of funds for emergency response measures when market shocks occur.

For effective budgeting:
1. **Contingency Planning**: Incorporate scenarios like the shock scenario into long-term financial planning.
2. **Reserve Fund Creation**: Develop and maintain a reserve fund that can be tapped during economic downturns caused by oil price volatility.

### Strategic Policy Recommendations to Improve 'Export Efficiency'

To enhance export efficiency, two key policy recommendations are:

1. **Investment in Technology and Infrastructure**:
   - **Upgrade Drilling Equipment**: Invest in modern drilling equipment and technology to increase production rates and reduce operational costs.
   - **Refinement Projects**: Develop or expand oil refining capabilities locally to add value to crude before exporting it. This can help in reducing dependency on exporting unrefined products, which often fetch lower prices.

2. **Streamlining Regulatory Processes**:
   - **Efficient Permitting System**: Simplify and expedite the permitting process for exploration and production activities.
   - **Harmonized Tax Regimes**: Align tax incentives and regulatory frameworks to encourage investment from both domestic and foreign companies.
   
By implementing these strategies, Nigeria can improve its export efficiency and mitigate some of the risks associated with structural decline and oil price volatility.
